```python
from abc import ABC, abstractmethod


class coup_de_ecriva(ABC):
   
    
    @abstractmethod
    def Fan_Page(self):
        pass


class FC_Cirok(coup_de_ecriva):
    """Subclass inheriting from coup_de_ecriva"""
    def Fan_Page(self):
        print("\nWelcome to FC Cirok!")


class Madiba_FC(coup_de_ecriva):
    """Subclass inheriting from coup_de_ecriva"""
    def Fan_Page(self):
        print("\nWelcome to Madiba FC!")


class Blue_Jay_FC(coup_de_ecriva):
    """Subclass inheriting from coup_de_ecriva"""
    def Fan_Page(self):
        print("\nWelcome to Blue Jay FC!")


class TSG_Walker(coup_de_ecriva):
    """Subclass inheriting from coup_de_ecriva"""
    def Fan_Page(self):
        print("\nWelcome to TSG Walker!")



print("--- Coup de Ecriva Fan Clubs ---")
print("1. FC Cirok")
print("2. Madiba FC")
print("3. Blue Jay FC")
print("4. TSG Walker")


user_choice = input("\nEnter the name of the club you support: ").strip().lower()


if user_choice == "fc cirok" or user_choice == "1":
    club_object = FC_Cirok()
    club_object.Fan_Page()

elif user_choice == "madiba fc" or user_choice == "2":
    club_object = Madiba_FC()
    club_object.Fan_Page()

elif user_choice == "blue jay fc" or user_choice == "3":
    club_object = Blue_Jay_FC()
    club_object.Fan_Page()

elif user_choice == "tsg walker" or user_choice == "4":
    club_object = TSG_Walker()
    club_object.Fan_Page()

else:
    print("\nInvalid choice! Please make sure you spelled the club name correctly.")
```
