```python
import pandas as pd
from abc import ABC

class StudentCouncil:
    def __init__(self, file_path):
        self.file_path = file_path
     
        self.__error_message = "Error: Student age exceeds 25. Access Denied."
        
    def get_error_message(self):
        """Getter method to return the encapsulated error message."""
        return self.__error_message

    def classify_students(self):
        
        try:
            df = pd.read_csv(self.file_path)
        except FileNotFoundError:
            print("SIS.csv not found.")
            return

       
        pirates = df[(df['Age'] > 14) & (df['Age'] < 18)]
        yankees = df[(df['Age'] > 18) & (df['Age'] < 22)]
        bulls = df[(df['Age'] > 22) & (df['Age'] < 25)]
        
       
        error_students = df[df['Age'] > 25]

      
        self._save_to_csv(pirates, "The_Pirates.csv")
        self._save_to_csv(yankees, "The_Yankees.csv")
        self._save_to_csv(bulls, "The_Bulls.csv")
        
      
        if not error_students.empty:
            print("\n--- Encapsulation Alerts ---")
            for name in error_students['Name']:
                print(f"Request for {name}: {self.get_error_message()}")

    def _save_to_csv(self, dataframe, filename):
        if not dataframe.empty:
            dataframe.to_csv(filename, index=False)
            print(f"\nCreated {filename}:")
            print(dataframe)


sis_manager = StudentCouncil('SIS.csv')
sis_manager.classify_students()
```
