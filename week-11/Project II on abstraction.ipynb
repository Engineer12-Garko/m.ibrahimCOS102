```python
import tkinter as tk
from abc import ABC, abstractmethod


class External_Vendors(ABC):
    """Abstract Base Class for PAU External Food Vendors."""
    @abstractmethod
    def menu(self):
        pass


class CooperativeCafeteria(External_Vendors):
    def menu(self):
        return (
            "Jollof Rice and Stew - N200\n"
            "White Rice and Stew - N200\n"
            "Fried Rice - N200\n"
            "Salad - N100\n"
            "Plantain - N100"
        )


class FaithHostelCafeteria(External_Vendors):
    def menu(self):
        return (
            "Fried Rice - N400\n"
            "White Rice and Stew - N400\n"
            "Jollof Rice - N400\n"
            "Beans - N200\n"
            "Chicken - N1000"
        )


class StudentCentreCafeteria(External_Vendors):
    def menu(self):
        return (
            "Chicken Fried Rice - N800\n"
            "Pomo Sauce - N300\n"
            "Spaghetti Jollof - N500\n"
            "Amala/Ewedu - N500\n"
            "Semo with Eforiro Soup - N500"
        )



def display_menu():
    # Get the value from the selected radio button
    choice = vendor_choice.get()
    
    if choice == 1:
        vendor = CooperativeCafeteria()
        output_label.config(text=vendor.menu())
    elif choice == 2:
        vendor = FaithHostelCafeteria()
        output_label.config(text=vendor.menu())
    elif choice == 3:
        vendor = StudentCentreCafeteria()
        output_label.config(text=vendor.menu())
    else:
        output_label.config(text="Please select a vendor first!")


root = tk.Tk()
root.title("PAU Vendors")
root.geometry("400x400")


instruction = tk.Label(root, text="Select an External Vendor:", font=("Arial", 12, "bold"))
instruction.pack(pady=10)


vendor_choice = tk.IntVar()


rb1 = tk.Radiobutton(root, text="Cooperative Cafeteria", variable=vendor_choice, value=1)
rb1.pack()

rb2 = tk.Radiobutton(root, text="Faith Hostel Cafeteria", variable=vendor_choice, value=2)
rb2.pack()

rb3 = tk.Radiobutton(root, text="Student Centre Cafeteria", variable=vendor_choice, value=3)
rb3.pack()


show_button = tk.Button(root, text="Show Menu", command=display_menu)
show_button.pack(pady=15)


menu_frame = tk.LabelFrame(root, text=" Vendor Menu ")
menu_frame.pack(pady=10, fill="both", expand=True, padx=20)

output_label = tk.Label(menu_frame, text="", font=("Arial", 11), justify="left")
output_label.pack(pady=10)


root.mainloop()
```
